## 1D

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Plotting the histogram as density and overlaying the normal distribution curve
plt.figure(figsize=(8, 5))

# Plot the density histogram
count, bins, ignored = plt.hist(df['Weight (kg)'], bins=8, density=True, color='skyblue', edgecolor='black', alpha=0.7, label='Data density')

# Calculate mean and standard deviation
mu, std = df['Weight (kg)'].mean(), df['Weight (kg)'].std()

# Generate points for the normal distribution curve
x_axis = np.linspace(df['Weight (kg)'].min() - 5, df['Weight (kg)'].max() + 5, 100)
plt.plot(x_axis, norm.pdf(x_axis, mu, std), color='red', linewidth=2, label='Normal Dist Fit')

plt.title('Histogram of Weight (kg) with Normal Curve')
plt.xlabel('Weight (kg)')
plt.ylabel('Density')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Extract variables
x = df['Height (m)']
y = df['Weight (kg)']

# Fit a linear trendline (y = a + bx -> y = intercept + slope * x)
# np.polyfit(x, y, 1) returns coefficients in decreasing power: [slope (b), intercept (a)]
b, a = np.polyfit(x, y, 1)

# Predict y values based on the model
y_pred = a + b * x

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y, y_pred))

# Plot the actual data points
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='skyblue', edgecolor='black', alpha=0.8, label='Data Points')

# Plot the trendline
plt.plot(x, y_pred, color='red', linewidth=2, label=f'Trendline: y = {a:.2f} + {b:.2f}x')

# Add text block with the metrics on the plot
text_stats = f'a (intercept) = {a:.4f}\nb (slope) = {b:.4f}\nRMSE = {rmse:.4f}'
plt.gca().text(0.05, 0.95, text_stats, transform=plt.gca().transAxes, fontsize=12,
             verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.title('Height vs. Weight with Regression Trendline')
plt.xlabel('Height (m)')
plt.ylabel('Weight (kg)')
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Display calculated values in text
print(f"Regression formula: y = {a:.4f} + {b:.4f} * x")
print(f"Slope (b): {b:.4f}")
print(f"Intercept (a): {a:.4f}")
print(f"RMSE: {rmse:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Split data by Gender
male_df = df[df['Gender'] == 'M']
female_df = df[df['Gender'] == 'F']

# --- MALE REGRESSION ---
x_m = male_df['Height (m)']
y_m = male_df['Weight (kg)']
b_m, a_m = np.polyfit(x_m, y_m, 1)
y_pred_m = a_m + b_m * x_m
rmse_m = np.sqrt(mean_squared_error(y_m, y_pred_m))

# --- FEMALE REGRESSION ---
x_f = female_df['Height (m)']
y_f = female_df['Weight (kg)']
b_f, a_f = np.polyfit(x_f, y_f, 1)
y_pred_f = a_f + b_f * x_f
rmse_f = np.sqrt(mean_squared_error(y_f, y_pred_f))

# --- PLOTTING ---
plt.figure(figsize=(10, 8))

# Scatter points
plt.scatter(x_m, y_m, color='skyblue', edgecolor='black', alpha=0.8, s=80, label='Male (M)')
plt.scatter(x_f, y_f, color='pink', edgecolor='black', alpha=0.8, s=80, label='Female (F)')

# Trendlines
# Sorting values for clean line plotting
sort_m = np.argsort(x_m)
plt.plot(x_m.iloc[sort_m], y_pred_m.iloc[sort_m], color='blue', linewidth=2,
         label=f'Male Trend: y = {a_m:.2f} + {b_m:.2f}x')

sort_f = np.argsort(x_f)
plt.plot(x_f.iloc[sort_f], y_pred_f.iloc[sort_f], color='deeppink', linewidth=2,
         label=f'Female Trend: y = {a_f:.2f} + {b_f:.2f}x')

# Add statistics text box on the plot
text_m = f'Male (M):\n  a = {a_m:.4f}\n  b = {b_m:.4f}\n  RMSE = {rmse_m:.4f}'
text_f = f'Female (F):\n  a = {a_f:.4f}\n  b = {b_f:.4f}\n  RMSE = {rmse_f:.4f}'
combined_text = f'{text_m}\n\n{text_f}'

plt.gca().text(0.05, 0.95, combined_text, transform=plt.gca().transAxes, fontsize=11,
             verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.title('Height vs. Weight with Gender-specific Trendlines')
plt.xlabel('Height (m)')
plt.ylabel('Weight (kg)')
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Display calculated values in console text
print("=== Male (M) Fit Details ===")
print(f"Equation: y = {a_m:.4f} + {b_m:.4f} * x")
print(f"RMSE: {rmse_m:.4f}\n")

print("=== Female (F) Fit Details ===")
print(f"Equation: y = {a_f:.4f} + {b_f:.4f} * x")
print(f"RMSE: {rmse_f:.4f}")
```
